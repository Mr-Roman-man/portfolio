## Popcorn — Checkpoint: Expand the catalog

**Prediction:** With 8 catalog entries, linear search may inspect all 8 in the worst case. Binary search needs at most 4 checks by the lesson's calculator. Doubling the catalog to 16 doubles linear checks to 16, while the binary estimate increases by one, to 5.

The binary-search estimate assumes a sorted catalog and direct access to its middle entry.

```python
# Popcorn checkpoint: worst-case inspection estimates
import math

for n in (8, 16):
    linear_checks = n
    binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
    print(f"Catalog size: {n}")
    print("Linear checks:", linear_checks)
    print("Binary checks:", binary_checks)
```
