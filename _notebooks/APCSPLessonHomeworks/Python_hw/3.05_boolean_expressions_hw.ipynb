## Popcorn Hack 1 — Check SFI Part Fields

**Prediction/result:** The supplied record has a product name, an accepted category, a spec number, and an effective date, so every individual check should print `True`. Changing the category to `Street Car` should make the category check `False`.

```python
# Popcorn Hack 1: Check SFI Part Fields
part = {
    "product_name": "Replacement Flywheels and Clutch Assemblies",
    "category": "Auto Racing",
    "spec_number": "1.1",
    "effective_date": "Nov. 9, 2001",
}
valid_categories = ["Auto Racing", "Drag Racing"]

has_product_name = bool(part["product_name"].strip())
has_valid_category = part["category"] in valid_categories
has_spec_number = bool(part["spec_number"].strip())
has_effective_date = bool(part["effective_date"].strip())

print("Product name present:", has_product_name)
print("Category accepted:", has_valid_category)
print("Spec number present:", has_spec_number)
print("Effective date present:", has_effective_date)
```

## Popcorn Hack 2 — Backend Create Rule

**Prediction/result:** The new `1.2` candidate should produce `True` (create). The candidate using existing spec `1.1` should produce `False` (reject), despite having all required fields.

```python
# Popcorn Hack 2: Backend create decision
valid_categories = ["Auto Racing", "Drag Racing"]
existing_spec_numbers = ["1.1", "2.1"]

def can_create_record(part):
    has_product_name = bool(part["product_name"].strip())
    has_valid_category = part["category"] in valid_categories
    has_spec_number = bool(part["spec_number"].strip())
    has_effective_date = bool(part["effective_date"].strip())
    is_new_spec = part["spec_number"] not in existing_spec_numbers
    return (
        has_product_name
        and has_valid_category
        and has_spec_number
        and has_effective_date
        and is_new_spec
    )

valid_candidate = {
    "product_name": "Multiple Disc Clutch Assemblies",
    "category": "Auto Racing",
    "spec_number": "1.2",
    "effective_date": "Feb. 9, 2006",
}
duplicate_candidate = {
    "product_name": "Existing Flywheel Record",
    "category": "Drag Racing",
    "spec_number": "1.1",
    "effective_date": "Jan. 1, 2026",
}

print("Valid candidate — create:", can_create_record(valid_candidate))
print("Duplicate candidate — create:", can_create_record(duplicate_candidate))
```

## Popcorn Hack 3 — SFI Search Filter

**Prediction/result:** With query `"1.2"`, only the matching Drag Racing record should appear. Its category is accepted, and the query exactly matches its spec number. A product-name query such as `"flywheel"` should match records whose product name contains that text, still excluding records outside the accepted categories.

```python
# CODE_RUNNER: Popcorn Hack 3 - SFI Search Filter
query = "1.2"
records = [
    {
        "product_name": "Replacement Flywheels and Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.1",
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2",
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1",
    },
]
valid_categories = ["Auto Racing", "Drag Racing"]

matches = []
for record in records:
    query_matches = (
        query.lower() in record["product_name"].lower()
        or query == record["spec_number"]
    )
    category_is_valid = record["category"] in valid_categories
    if query_matches and category_is_valid:
        matches.append(record)

print("Matching SFI records:")
for record in matches:
    print(record["product_name"], "—", record["spec_number"])
```

## Independent Homework — SFI Backend Validator

**Expected decisions:** The Multiple Disc record should be accepted because its fields are complete, its category is allowed, and spec `1.2` is new. The Street Car record should be rejected for its category. The Existing Flywheel record should be rejected because spec `1.1` is already stored. The extra incomplete record should be rejected because its effective date is empty.

The validator evaluates each requirement separately, then combines them with `and`. `not` is applied to the duplicate check so only new spec numbers pass.

```python
# Homework Hack: SFI Backend Validator
existing_spec_numbers = ["1.1", "2.1"]
valid_categories = ["Auto Racing", "Drag Racing"]

test_records = [
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.2",
        "effective_date": "Feb. 9, 2006",
    },
    {
        "product_name": "Replacement Flywheels",
        "category": "Street Car",
        "spec_number": "3.1",
        "effective_date": "Jan. 1, 2026",
    },
    {
        "product_name": "Existing Flywheel Record",
        "category": "Drag Racing",
        "spec_number": "1.1",
        "effective_date": "Jan. 1, 2026",
    },
    {
        "product_name": "Incomplete Clutch Record",
        "category": "Drag Racing",
        "spec_number": "4.1",
        "effective_date": "",
    },
]

def validate_record(record):
    """Return True only when the record meets every create requirement."""
    has_product_name = bool(record["product_name"].strip())
    has_valid_category = record["category"] in valid_categories
    has_spec_number = bool(record["spec_number"].strip())
    has_effective_date = bool(record["effective_date"].strip())
    spec_is_new = record["spec_number"] not in existing_spec_numbers

    return (
        has_product_name
        and has_valid_category
        and has_spec_number
        and has_effective_date
        and spec_is_new
    )

for candidate in test_records:
    accepted = validate_record(candidate)
    decision = "ACCEPT" if accepted else "REJECT"
    print(f"{decision}: {candidate['product_name']} (spec {candidate['spec_number']})")
```
