## Popcorn Hack 1 — After-School Homework Plan

**Prediction for the starter values:** The outer `if is_weekday` branch is taken, then the inner `else` branch prints `Finish homework`. The tests below also reach the other two outcomes.

````python
# CODE_RUNNER: Popcorn Hack 1 - test the after-school homework plan
def after_school_plan(is_weekday, homework_finished):
    if is_weekday:
        if homework_finished:
            print("Free time")
        else:
            print("Finish homework")
    else:
        print("Weekend plan")

# Starter values, then the other two outcomes.
print("Starter values:")
after_school_plan(True, False)
print("Weekday, homework finished:")
after_school_plan(True, True)
print("Weekend:")
after_school_plan(False, False)
````

## Popcorn Hack 2 — Repair the Student Account Check

In the starter, the permission `if` is aligned with `if logged_in`, so it runs independently—even when no user is logged in. Nest the permission check inside the login branch, and give the outer login check its own `else`.

````python
# CODE_RUNNER: Popcorn Hack 2 - nest the permission check after login
def account_check(logged_in, has_permission):
    if logged_in:
        print("Account found")
        if has_permission:
            print("Page opened")
        else:
            print("Permission needed")
    else:
        print("Log in first")

print("Logged in with permission:")
account_check(True, True)
print("Logged in without permission:")
account_check(True, False)
print("Not logged in:")
account_check(False, False)
````

## Popcorn Hack 3 — Lunch and Sports Practice

**Condition order:** First, check whether lunch is packed. Only inside that branch do we check whether there is after-school practice; the same practice check is nested in the other lunch branch. The four combinations each print a different recommendation.

JavaScript equivalent:

```javascript
if (lunchPacked) {
  if (practiceAfterSchool) {
    console.log("Packed lunch; bring an extra snack for practice.");
  } else {
    console.log("Packed lunch; no extra practice snack needed.");
  }
} else {
  if (practiceAfterSchool) {
    console.log("Buy lunch; bring an extra snack for practice.");
  } else {
    console.log("Buy lunch; no extra practice snack needed.");
  }
}
```

College Board pseudocode uses the same nesting (with 1-based syntax conventions irrelevant to Booleans):

```text
IF (lunchPacked)
{
    IF (practiceAfterSchool)
    {
        DISPLAY("Packed lunch; bring an extra snack for practice.")
    }
    ELSE
    {
        DISPLAY("Packed lunch; no extra practice snack needed.")
    }
}
ELSE
{
    IF (practiceAfterSchool)
    {
        DISPLAY("Buy lunch; bring an extra snack for practice.")
    }
    ELSE
    {
        DISPLAY("Buy lunch; no extra practice snack needed.")
    }
}
```

````python
# CODE_RUNNER: Popcorn Hack 3 - test all lunch and practice paths
def lunch_and_practice_plan(lunch_packed, practice_after_school):
    if lunch_packed:
        if practice_after_school:
            print("Packed lunch; bring an extra snack for practice.")
        else:
            print("Packed lunch; no extra practice snack needed.")
    else:
        if practice_after_school:
            print("Buy lunch; bring an extra snack for practice.")
        else:
            print("Buy lunch; no extra practice snack needed.")

lunch_and_practice_plan(True, True)
lunch_and_practice_plan(True, False)
lunch_and_practice_plan(False, True)
lunch_and_practice_plan(False, False)
````

## Homework Hack — High School Daily Routine Dashboard

The outer login decision must succeed before the dashboard reads the student's schedule. Inside the logged-in branch, the program checks homework status and then nests a practice decision inside the lunch choice. This keeps each recommendation tied to the correct account and schedule path.

````python
# CODE_RUNNER: Homework Hack - High School Daily Routine Dashboard
student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True,
}

def daily_routine(student):
    if student["logged_in"]:
        print("Account access granted.")

        if student["homework_finished"]:
            print("Homework is finished — free time is available.")
        else:
            print("Finish homework before free time.")

        if student["lunch_packed"]:
            if student["practice_today"]:
                print("Lunch is packed; bring an extra snack for practice.")
            else:
                print("Lunch is packed; no after-school practice snack is needed.")
        else:
            if student["practice_today"]:
                print("Plan to buy lunch and bring an extra snack for practice.")
            else:
                print("Plan to buy lunch; there is no after-school practice today.")
    else:
        print("Log in first to view your daily plan.")

# Run the supplied starter data.
daily_routine(student)
````

````python
# CODE_RUNNER: Homework tests - exercise every account and schedule path
print("Test: logged out")
daily_routine({
    "logged_in": False,
    "homework_finished": False,
    "lunch_packed": False,
    "practice_today": False,
})

for homework_finished in [False, True]:
    for lunch_packed in [False, True]:
        for practice_today in [False, True]:
            print(
                "Test: logged in | homework finished:", homework_finished,
                "| lunch packed:", lunch_packed,
                "| practice:", practice_today,
            )
            daily_routine({
                "logged_in": True,
                "homework_finished": homework_finished,
                "lunch_packed": lunch_packed,
                "practice_today": practice_today,
            })
````
