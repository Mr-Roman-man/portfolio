## Design Thinking — Player-centered game settings

- **Player need:** A player may prefer reduced motion for comfort and accessibility.
- **Goal:** Let the player make a clear choice and preserve understandable game progress.
- **Representation considered:** Store the preference as a Boolean (`True` or `False`) and keep each level's star count as an integer.
- **Prototype:** The profile and star-counter code above stores the game name, reduced-motion setting, separate level scores, and updated total.
- **Test/revision:** Traced both a 5-star bonus and a zero-star bonus from fresh starting values. The test shows that the saved total is a snapshot and must be recalculated after a level score changes.

Reduced motion should be a player choice because players have different comfort and access needs; the setting should not be assumed or forced for everyone.
