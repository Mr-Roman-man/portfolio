## Popcorn Hack 2 — Write and test Flask routes

**Extra route added:** `/coach`, which returns a JSON object with the coach name and home/away records. The test client requests `/score`, `/roster`, and `/coach` without requiring a browser.

````python
# CODE_RUNNER: Popcorn Hack 2 - test the basketball stats Flask routes
from flask import Flask, jsonify

app = Flask(__name__)

# Return the current home and away score from the score route.
@app.route("/score")
def score():
    return jsonify({"home": 58, "away": 52, "quarter": 3})

# Return the team roster from the roster route.
@app.route("/roster")
def roster():
    return jsonify({"players": ["Maya", "Jordan", "Alex"]})

# Add a custom route with coach and season-record information.
@app.route("/coach")
def coach():
    return jsonify({"coach": "Coach Lee", "wins": 8, "losses": 2})

# Use Flask's test client to call each route without a browser.
client = app.test_client()
print("/score says:", client.get("/score").get_json())
print("/roster says:", client.get("/roster").get_json())
print("/coach says:", client.get("/coach").get_json())
````

## Homework — Season Stats Report

This version uses fictional basketball game data. Its documented `summarize_season()` procedure returns the total and average points, best game, and points scored against each opponent. It also uses `datetime` to calculate the days until the next game and `json` to serialize the report.

````python
# CODE_RUNNER: Homework - basketball season stats report with standard libraries
import json
import statistics
from datetime import date

games = [
    {"date": "2026-10-02", "opponent": "Harbor High", "points": 68, "minutes": 30},
    {"date": "2026-10-06", "opponent": "Canyon High", "points": 61, "minutes": 28},
    {"date": "2026-10-11", "opponent": "Harbor High", "points": 75, "minutes": 32},
    {"date": "2026-10-18", "opponent": "Mesa High", "points": 70, "minutes": 29},
]


def summarize_season(season):
    """Summarize fictional basketball season results.

    Parameters:
        season: A non-empty list of game dictionaries. Each dictionary has
            date, opponent, points, and minutes fields.

    Returns:
        A dictionary containing total_points, average_points, best_game,
        and points_by_opponent.
    """
    point_scores = []
    points_by_opponent = {}

    for game in season:
        point_scores.append(game["points"])
        opponent = game["opponent"]
        if opponent in points_by_opponent:
            points_by_opponent[opponent] = points_by_opponent[opponent] + game["points"]
        else:
            points_by_opponent[opponent] = game["points"]

    best_game = max(season, key=lambda game: game["points"])
    return {
        "total_points": sum(point_scores),
        "average_points": statistics.mean(point_scores),
        "best_game": best_game,
        "points_by_opponent": points_by_opponent,
    }


# Summarize the season, then add a date-based next-game countdown.
season_report = summarize_season(games)
next_game_date = date(2026, 11, 12)
days_until_next_game = (next_game_date - date.today()).days
season_report["next_game_date"] = next_game_date.isoformat()
season_report["days_until_next_game"] = days_until_next_game

# Serialize and display the report as JSON text.
print(json.dumps(season_report, indent=2))
print("Days until next game:", days_until_next_game)
````
