## Popcorn Hack 3 — Decompose a long procedure and test its parts

**Task:** Break one long dessert-making procedure into smaller procedures. Each helper can be tested independently, and the main procedure combines the steps. This cell tests each helper and then the complete recipe.

````python
# CODE_RUNNER: Popcorn Hack 3 - decompose and test dessert steps
def measure_flour_for_test(cups):
    print(f"Measuring {cups} cups of flour.")
    return cups


def mix_ingredients_for_test(flour_cups, sugar_cups, eggs):
    print(f"Mixing {flour_cups} cups flour, {sugar_cups} cups sugar, and {eggs} eggs.")
    return True


def bake_dessert_for_test(temperature, minutes):
    print(f"Baking at {temperature} degrees for {minutes} minutes.")
    return temperature, minutes


def make_decomposed_dessert():
    flour_cups = measure_flour_for_test(2)
    mixed = mix_ingredients_for_test(flour_cups, 1, 2)
    bake_settings = bake_dessert_for_test(350, 25)
    if mixed and bake_settings == (350, 25):
        print("Dessert is ready!")


# Test each procedure separately, then test the complete recipe.
print("Testing flour measurement:")
measure_flour_for_test(2)
print("Testing mixing step:")
mix_ingredients_for_test(2, 1, 2)
print("Testing bake step:")
bake_dessert_for_test(350, 25)
print("Testing complete recipe:")
make_decomposed_dessert()
````

## Homework — Build a reusable dessert procedure

Write a `prepare_dessert(servings)` procedure that coordinates small helper procedures for ingredient amounts, mixing, and baking. Each helper has one clear task; the servings parameter lets the recipe be reused. Test both a small and a larger batch, plus an invalid zero-serving case.

````python
# CODE_RUNNER: Homework - reusable dessert procedure with helper tasks
def calculate_ingredient_amount(amount_per_serving, servings):
    return amount_per_serving * servings


def list_dessert_ingredients(servings):
    flour_cups = calculate_ingredient_amount(0.5, servings)
    sugar_cups = calculate_ingredient_amount(0.25, servings)
    eggs = calculate_ingredient_amount(0.5, servings)
    print(f"Ingredients for {servings} servings:")
    print(f"Flour: {flour_cups} cups")
    print(f"Sugar: {sugar_cups} cups")
    print(f"Eggs: {eggs}")


def mix_homework_ingredients():
    print("Mix the ingredients until combined.")


def bake_homework_dessert():
    print("Bake at 350 degrees for 25 minutes.")


def prepare_dessert(servings):
    if servings <= 0:
        print("Servings must be greater than zero.")
    else:
        list_dessert_ingredients(servings)
        mix_homework_ingredients()
        bake_homework_dessert()
        print("Dessert is ready!")


# Test a small batch, a larger batch, and invalid input.
prepare_dessert(2)
prepare_dessert(4)
prepare_dessert(0)
````
