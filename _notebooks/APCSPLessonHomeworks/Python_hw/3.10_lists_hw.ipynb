## Popcorn Hack 2 — Sum and average volunteer hours

**Prediction:** The six volunteers worked 42 hours total, for an average of 7.0 hours. The empty-list test should print `No data yet` rather than divide by zero.

```python
# CODE_RUNNER: Lists Popcorn 2 - volunteer-hours sum and average
volunteer_hours = [7, 6, 8, 5, 7, 9]

# Initialize the accumulator before traversing the list.
total_hours = 0
for hours in volunteer_hours:
    total_hours = total_hours + hours

# Calculate the average only when the list contains data.
if len(volunteer_hours) == 0:
    print("No data yet")
else:
    average_hours = total_hours / len(volunteer_hours)
    print("Total hours:", total_hours)
    print("Average hours:", average_hours)

# Test the empty-list boundary without dividing by zero.
empty_hours = []
if len(empty_hours) == 0:
    print("No data yet")
else:
    empty_total = 0
    for hours in empty_hours:
        empty_total = empty_total + hours
    print("Average hours:", empty_total / len(empty_hours))
```

## Popcorn Hack 3 — Count and filter donation requests

**Prediction:** The frequency dictionary should show blankets: 1, socks: 3, meals: 2, and water: 1. The items with at least two votes are `socks` and `meals`.

```python
# CODE_RUNNER: Lists Popcorn 3 - count then filter
items_requested = ["blankets", "socks", "meals", "socks", "water", "socks", "meals"]

# Count how many times each requested item appears.
frequency = {}
for item in items_requested:
    if item in frequency:
        frequency[item] = frequency[item] + 1
    else:
        frequency[item] = 1

# Collect each item that received at least two votes.
popular = []
for item in frequency:
    if frequency[item] >= 2:
        popular.append(item)

# Display the filtered list.
print(frequency)
print(popular)
```

## Homework — Donation inventory analysis

**Prediction:** There are 12 donations; the last item is `socks`. The costs total 31, so the integer average is 2. The frequency counts are blankets: 5, socks: 4, water: 2, meals: 1. Items donated at least three times are blankets and socks.

```python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter
donations_by_item = [
    "blankets", "socks", "blankets", "water", "socks", "blankets",
    "meals", "socks", "blankets", "water", "blankets", "socks",
]
costs = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# Lesson 1: print the number of donations and last donation using indexing.
print(len(donations_by_item))
print(donations_by_item[-1])

# Lesson 2: total all costs with an accumulator loop.
total = 0
for cost in costs:
    total = total + cost

# Print the total and integer average as requested.
if len(costs) == 0:
    print("No cost data")
else:
    print("Total:", int(total))
    print("Average:", int(total / len(costs)))

# Lesson 3: build a frequency dictionary for donation item names.
frequency = {}
for item in donations_by_item:
    if item in frequency:
        frequency[item] = frequency[item] + 1
    else:
        frequency[item] = 1
print(frequency)

# Filter to a new list of items donated at least three times.
frequent_items = []
for item in frequency:
    if frequency[item] >= 3:
        frequent_items.append(item)
print(frequent_items)
```
