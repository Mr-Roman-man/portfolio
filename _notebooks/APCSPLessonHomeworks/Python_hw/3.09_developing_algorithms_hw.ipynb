## Popcorn Hack 1 — Translate pseudocode to Python

**Prediction:** The four volunteer-hour values add to 14. Dividing by the four entries gives an average of 3.5 hours. The loop computes the total without using `sum()`.

```python
# CODE_RUNNER: Popcorn Hack 1 - translate hours algorithm to Python
# Step 1: Store the volunteer hours in a list.
hours = [3, 5, 2, 4]

# Step 2: Initialize the running total before the loop.
total = 0

# Step 3: Visit each hour value and add it to the total.
for h in hours:
    total = total + h

# Step 4: Divide the total by the number of volunteers to get the average.
average = total / len(hours)

# Step 5: Display the total and average.
print(total)
print(average)
```

## Popcorn Hack 2 — Fix the full-workshop counter

**Prediction before fix:** The starter condition `count > capacity` counts only 15, so it prints 1. This is a boundary bug: Self-Defense has exactly 12 of 12 seats filled, so the correct comparison is `>=`. With that correction, the answer is 2.

```python
# CODE_RUNNER: Popcorn Hack 2 - count workshops at or above capacity
signed_up = [12, 7, 15, 9]  # Self-Defense, Parenting Skills, Fitness, Self-Sufficiency
capacity = 12

full = 0
for count in signed_up:
    # Include workshops exactly at capacity as well as over capacity.
    if count >= capacity:
        full = full + 1

print("Full workshops:", full)
```

## Popcorn Hack 3 — Collect workshops with open seats

**Prediction:** Parenting Skills (7 signed up) and Self-Sufficiency (9 signed up) have fewer than 12 attendees. The collector should print 2 and those two names; Self-Defense is exactly at capacity, so it is not included.

```python
# CODE_RUNNER: Popcorn Hack 3 - collect workshops with open seats
workshops = ["Self-Defense", "Parenting Skills", "Fitness", "Self-Sufficiency"]
signed_up = [12, 7, 15, 9]  # Each count matches the workshop at the same index.
capacity = 12

# Start with an empty collection for workshops that still have seats.
open_workshops = []

# Check each aligned workshop and signup count.
for i in range(len(workshops)):
    if signed_up[i] < capacity:
        open_workshops.append(workshops[i])

# Report the number of open workshops and their names.
print("Workshops with open seats:", len(open_workshops))
print(open_workshops)
```

## Homework Hack — Safe Passage Heals supply restock report

**Prediction:** Blankets need 12 and Shampoo needs 10. Socks are exactly at their target of 25, so they are not short. The total shortage is 22, and Blankets are the most urgent item with a shortage of 12.

```python
# CODE_RUNNER: 3.09 Homework - Safe Passage Heals supply restock report
items = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]  # Supplies currently in the closet.
target = [30, 20, 25, 15, 30]  # Desired amount for each item.

# Step 1: Collect each item below its target and the shortage amount.
short_items = []
short_amounts = []
for i in range(len(items)):
    # Add an item and its shortage only when stock is below the target.
    if in_stock[i] < target[i]:
        short_items.append(items[i])
        short_amounts.append(target[i] - in_stock[i])

# Step 2: Sum shortages with a loop instead of sum().
total_needed = 0
for shortage in short_amounts:
    total_needed = total_needed + shortage

# Step 3: Find the largest shortage and keep its index for the item name.
most_urgent_index = 0
for i in range(1, len(short_amounts)):
    # Keep the index of the largest shortage seen so far.
    if short_amounts[i] > short_amounts[most_urgent_index]:
        most_urgent_index = i

# Step 4: Print the appropriate report, including the no-shortage case.
if len(short_items) == 0:
    print("Fully stocked!")
else:
    print("Restock needed:")
    # Print every item and its shortage amount.
    for i in range(len(short_items)):
        print(f"{short_items[i]}: need {short_amounts[i]}")
    print("Total items needed:", total_needed)
    print(
        f"Most urgent: {short_items[most_urgent_index]} "
        f"(short by {short_amounts[most_urgent_index]})"
    )
```
