## Homework — Grow your UESL game

**Prediction:** Represent the original three levels as `[12, 18, 10]`. Add 5 to the second level and append a fourth level with 7 stars; expect `[12, 23, 10, 7]`, 4 levels, and 52 total stars. The collection keeps each level in order, and one traversal can total any list length.

```python
# Homework: UESL level collection and star total
level_stars = [12, 18, 10]

# Add a five-star bonus to the second level if it exists.
if len(level_stars) > 1:
    level_stars[1] = level_stars[1] + 5

# Add a new level, then total all levels with one traversal.
level_stars.append(7)
total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print("Level stars:", level_stars)
print("Number of levels:", len(level_stars))
print("Total stars:", total_stars)
```

## Design Thinking — Manage levels with a collection

- **Need:** A game maker needs to add or reorder levels without maintaining a separate variable and total formula for every level.
- **Goal:** Keep star counts in level order and calculate totals reliably as the game grows.
- **Representation considered:** Separate variables are straightforward for a few levels but require more code edits for each added level; a list stores all counts in one ordered collection.
- **Prototype:** The program updates the second level, appends a new level, and traverses the list to calculate the total.
- **Test/revision:** Tested a populated four-level game and an empty collection. The empty test skips the second-level update and resets the total to zero before traversal, so it handles no levels safely.
- **Complexity:** With a list and a loop, adding a level requires one append rather than a new variable and another term in a hard-coded sum. The traversal still processes each level once.
- **Player-facing design:** Use clear level labels in a game interface so players do not need to understand zero-based indices.

Changing a list does not automatically recalculate a previously stored total; run the traversal again after updating the collection.
