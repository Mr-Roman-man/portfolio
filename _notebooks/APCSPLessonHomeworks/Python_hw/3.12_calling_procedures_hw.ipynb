## Homework Hack: Emergency Response Dispatch

`check_priority()` returns a Boolean, and `dispatch_response()` calls it to choose a response. The examples below use three different incident types and multiple locations.

```python
# Homework Hack: Emergency Response Dispatch
def check_priority(incident_type):
    """Return whether an incident requires priority handling."""
    return incident_type in {"Medical Emergency", "Fire"}


def dispatch_response(location, incident_type):
    """Select and print a response using the priority helper."""
    priority = check_priority(incident_type)
    if priority:
        print(f"PRIORITY RESPONSE: Dispatch emergency services for {incident_type} in {location}.")
    else:
        print(f"STANDARD RESPONSE: Record {incident_type} in {location} for follow-up.")


dispatch_response("Poway", "Medical Emergency")
dispatch_response("Ramona", "Power Outage")
dispatch_response("Escondido", "Fire")
```
