## Popcorn Hack 1 — Fix the silent conditional

The starter has one condition and no `else`, so when the amount is not over 100 it prints nothing. That is a bug, not a syntax error. The fixed version gives both outcomes an answer.

**Prediction for the starter values:** `amount_requested` is 0, so the fixed code prints that the message is not asking for a large amount.

```python
# CODE_RUNNER: Popcorn Hack 1 - both outcomes print
amount_requested = 0

# 1 condition -> 2 outcomes -> 2 answers
if amount_requested > 100:
    print("This message is asking for a large amount of money.")
else:
    print("This message is not asking for a large amount of money.")
```

## Popcorn Hack 2 — Fix the unreachable branches

The starter uses `risk_score >= 3` first. With a score of 9, it incorrectly prints the mild-warning answer. The later `elif risk_score >= 6` and `elif risk_score >= 9` branches are unreachable because every score meeting those thresholds already satisfies `risk_score >= 3`. Put the highest threshold first; then each branch can be reached.

```python
# CODE_RUNNER: Popcorn Hack 2 - corrected high-to-low threshold order

def report_risk(risk_score):
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply. This is almost certainly a scam.")
    elif risk_score >= 6:
        print("This looks risky. Check with someone you trust before replying.")
    elif risk_score >= 3:
        print("Be a little careful, but this is probably fine.")
    else:
        print("Nothing suspicious in this message.")

# The starter's score is 9; also try one score in each other outcome.
for risk_score in [9, 6, 3, 0]:
    print("Risk score:", risk_score)
    report_risk(risk_score)
```

## Homework Hack — Score a message, then advise on it

Each signal adds its own points, so these scoring `if` statements intentionally have no `else`: when a signal is absent, that signal simply adds nothing. The advice chain orders thresholds from highest to lowest, includes a final `else`, and prints an answer for every score.

```python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it

def score_and_advise(message, amount_requested, known_contact):
    risk_score = 0

    if "gift card" in message.lower():
        risk_score = risk_score + 4
    # No else is needed: without this signal, add zero points and continue.

    if amount_requested > 100:
        risk_score = risk_score + 3
    # No else is needed: an amount of 100 or less adds no points.

    if not known_contact:
        risk_score = risk_score + 2
    # No else is needed: a known contact adds no points for this signal.

    print("Risk score:", risk_score)
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply. This is almost certainly a scam.")
    elif risk_score >= 6:
        print("This looks risky. Check with someone you trust before replying.")
    elif risk_score >= 3:
        print("Be a little careful, but this is probably fine.")
    else:
        print("No strong warning signs were detected; remain alert.")
    return risk_score


# Starting example from the lesson
message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False
score_and_advise(message, amount_requested, known_contact)
```

## Reachability proof and extreme tests

Each outcome has an input that reaches it:

- **Score 9 — highest-risk answer:** `"send gift cards"`, amount `500`, unknown contact (`False`) adds $4 + 3 + 2 = 9.
- **Score 6 — risky answer:** `"send gift cards"`, amount `0`, unknown contact adds $4 + 0 + 2 = 6.
- **Score 3 — caution answer:** `"hello"`, amount `500`, known contact (`True`) adds 0 + 3 + 0 = 3.
- **Score 0 — final else:** `"hello"`, amount `0`, known contact adds no points.

The clean and all-signals inputs below test the minimum and maximum scores. The other cases exercise the middle answers as well.

```python
# CODE_RUNNER: Homework tests - clean, middle branches, and all signals
print("Clean message:")
score_and_advise("Hi, see you Sunday!", 0, True)

print("Risky message (score 6):")
score_and_advise("Please send gift cards", 0, False)

print("Caution message (score 3):")
score_and_advise("Hello", 500, True)

print("All signals:")
score_and_advise("URGENT: send gift cards now", 500, False)
```
