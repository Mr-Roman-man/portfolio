# 3.08 Iterations — Popcorn Hack and Homework

Completed the SFSRC wildfire dispatch scanner and a reusable themed loop-based monitoring checker.

## Popcorn Hack — SFSRC wildfire dispatch scanner

**Prediction:** Alerts below 3 are skipped. The scanner should count the action alerts and total their levels through the first critical alert (8 or higher), then stop without processing later entries.

```python
# CODE_RUNNER: Popcorn Hack - wildfire dispatch scanner
# Scan SFSRC wildfire alert levels, skip safe zones, and stop at a critical alert.
alerts = [1, 4, 2, 6, 3, 9, 5]
safe_cutoff = 3
critical_cutoff = 8

action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

for alert in alerts:
    if alert < safe_cutoff:
        safe_count = safe_count + 1
        print("Safe alert skipped:", alert)
        continue

    action_count = action_count + 1
    total_action_score = total_action_score + alert
    print("Dispatch responding to alert:", alert)

    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found:", alert)
        break

print("Safe alerts skipped:", safe_count)
print("Action alerts processed:", action_count)
print("Total action score:", total_action_score)
print("Critical alert found:", critical_found)
```

## Homework Hack — Themed loop-based checker

This version monitors SFSRC wildfire zone alert levels. The `for` loop checks at least eight readings, `continue` skips safe alerts, and `break` stops the scan at the first critical alert. It reports checked, skipped, action, total, and critical-found values.

```python
# CODE_RUNNER: Homework Hack - SFSRC wildfire monitoring checker

def scan_wildfire_alerts(alert_levels, safe_cutoff, critical_cutoff):
    checked_count = 0
    action_count = 0
    skipped_count = 0
    total_score = 0
    critical_found = False

    for alert in alert_levels:
        checked_count = checked_count + 1

        if alert < safe_cutoff:
            skipped_count = skipped_count + 1
            print("Safe zone skipped:", alert)
            continue

        action_count = action_count + 1
        total_score = total_score + alert
        print("Response needed for alert:", alert)

        if alert >= critical_cutoff:
            critical_found = True
            print("Critical alert reached; stopping scan.")
            break

    print("Final checked count:", checked_count)
    print("Final action count:", action_count)
    print("Final skipped count:", skipped_count)
    print("Final total action score:", total_score)
    print("Critical found:", critical_found)


# Nine zones; the scan stops at the first critical alert.
zone_alert_levels = [1, 4, 2, 6, 3, 9, 5, 7, 2]
safe_cutoff = 3
critical_cutoff = 8
scan_wildfire_alerts(zone_alert_levels, safe_cutoff, critical_cutoff)
```

## Additional test — no critical alert

This test confirms the loop completes the full list when no alert reaches the critical threshold. With cutoffs 3 and 8, the expected summary is 8 checked, 3 skipped, 5 action alerts, total score 25, and `Critical found: False`.

```python
# CODE_RUNNER: Homework test - complete scan without a critical alert
zone_alert_levels = [0, 1, 2, 3, 4, 5, 6, 7]
safe_cutoff = 3
critical_cutoff = 8
scan_wildfire_alerts(zone_alert_levels, safe_cutoff, critical_cutoff)
```
