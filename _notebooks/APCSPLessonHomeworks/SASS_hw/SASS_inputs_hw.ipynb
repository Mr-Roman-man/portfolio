# SASS Inputs — Popcorn Hack and Homework

Completed the base-input refactor and the PVO signup form using OCS input grammar, size modifiers, and accessible labels.

## Popcorn Hack — Refactor the PVO Inputs

**Completed:** The resource search is a large gradient OCS input; the volunteer first-name field is a compact small OCS input. Both use `ocs__input`, with no inline styles or invented classes.

```python
%%html
<!-- UI_RUNNER: Inputs Popcorn Base -->
<label for="pvo-resource-search">Search PVO resources</label>
<input id="pvo-resource-search" name="resourceSearch" type="text" class="ocs__input large gradient" placeholder="Search PVO resources...">

<label for="pvo-volunteer-name">Volunteer first name</label>
<input id="pvo-volunteer-name" name="volunteerFirstName" type="text" class="ocs__input small" placeholder="Volunteer first name">
```
