# OCS Grid Grammar — Popcorn Hacks and Homework

Completed the two grid Popcorn Hacks and the standard-grid plus calculator-grid homework using OCS classes only.

## Popcorn Hack 1 — Fix the Beaker Row

**Prediction:** The three beakers will render as cells in one standard OCS grid. The markup contains no inline layout style and no made-up class.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 - OCS standard beaker row -->
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell">Beaker A: Water</div>
  <div class="ocs__grid-cell">Beaker B: Salt Water</div>
  <div class="ocs__grid-cell">Beaker C: Vinegar</div>
</div>
```

## Popcorn Hack 2 — Size the Beakers with `fr`

**Prediction:** Four `fr` values create four columns. The Control Beaker receives 2 of the 5 total shares, so it is twice as wide as each other beaker (1 share each). The `style` value is only used here for the lesson's `fr` demonstration.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 - four fractional columns, control gets 2fr -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
  <div class="ocs__grid-cell">Beaker 1</div>
  <div class="ocs__grid-cell">Beaker 2</div>
  <div class="ocs__grid-cell">Beaker 3</div>
  <div class="ocs__grid-cell">Control Beaker</div>
</div>
```
