# SASS Button Grammar — Popcorn Hacks and Homework

Completed the base button, modifier/group, and semantic tone Popcorn Hacks, plus both My Good Brain action toolbar homeworks using OCS classes only.

## Popcorn Hack 1 — Add the Base Button Class

**Completed:** Added the `ocs__btn` base utility class to the semantic resource link.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 - add OCS base button class -->
<a class="ocs__btn" href="https://www.mygoodbrain.org/resources">Mental Health Resource Directory</a>
```

## Popcorn Hack 2 — Modifiers and Button Groups

**Completed:** Wrapped both actions in `ocs__links`. The primary check-in button uses the requested `pill` and `fill` modifiers; the draft action remains outlined.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 - group modified OCS buttons -->
<div class="ocs__links">
  <button class="ocs__btn accent pill fill">Complete Daily Check-In</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">Save Draft Journal</a>
</div>
```

## Popcorn Hack 3 — Status Tones and Wide Layout Bar

**Completed:** Changed the check-in tone to green, upgraded the grouping to `ocs__links--wide`, and gave the crisis-support link a filled red priority style.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 3 - wide semantic status action bar -->
<div class="ocs__links--wide">
  <button class="ocs__btn alert-yellow outline">Complete Check-In</button>
  <a class="ocs__btn alert-green outline" href="https://www.mygoodbrain.org/resources">Explore Self-Care Resources</a>
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources">Get Crisis Support</a>
</div>
```

## Homework Hack 1 — Adaptive Resource Toolbar

**Checklist: PASS** — The three semantic links are grouped by `ocs__links`; urgent support, self-care, and community actions use their specified OCS tone and shape modifiers. No inline styles or custom classes remain.

```python
%%html
<!-- UI_RUNNER: Homework Hack 1 - adaptive mental health resource toolbar -->
<div class="ocs__links">
  <a class="ocs__btn alert-red fill pill" href="https://www.mygoodbrain.org/resources">Urgent Support</a>
  <a class="ocs__btn alert-green fill pill" href="https://www.mygoodbrain.org/resources">Self-Care</a>
  <a class="ocs__btn accent outline pill" href="https://www.mygoodbrain.org/resources">Community</a>
</div>
```

## Homework Hack 2 — Primary vs. Secondary Action Priority

**Checklist: PASS** — The main mood-entry action uses a filled accent button; weekly trends is outlined; crisis support is an outlined red alert. All three actions are grouped in `ocs__links` with no inline styles or custom CSS classes.

```python
%%html
<!-- UI_RUNNER: Homework Hack 2 - primary, secondary, and crisis actions -->
<div class="ocs__links">
  <button class="ocs__btn accent fill">Log Mood Entry</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">View Weekly Trends</a>
  <a class="ocs__btn alert-red outline" href="https://www.mygoodbrain.org/resources">Crisis Hotline</a>
</div>
```
