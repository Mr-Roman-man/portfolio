# SASS Toggles — Popcorn Hacks and Homework

Completed the toggle vocabulary check, repaired the missing OCS class, customized a toggle color, and built an interactive three-setting PVO toggle panel.

## Popcorn Hack — Toggle Vocabulary Check

1. The **checkbox input** keeps track of whether a toggle is on or off.
2. The **toggle track** is the visible part that changes when the toggle is switched.
3. The `.switch` class controls the toggle's position, size, padding, and **color**.
4. The **`border-radius`** property gives the slider rounded edges.
5. The `&:before` pseudo-element controls the **position/movement of the toggle knob**.

## Find the Bug — Restore the Missing Toggle Track

The broken starter omits `ocs__toggle-track` from the decorative `<span>`. Add that class so the SASS can render and respond to the checkbox state.

```python
%%html
<!-- UI_RUNNER: Toggle Bug Fix - add the missing track class -->
<label class="ocs__toggle ocs__toggle--green">
  <input id="notifications-bugfix" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Enable Notifications</span>
</label>
<p id="notifications-bugfix-status" aria-live="polite">Notifications are off.</p>
<script>
  // The checked checkbox is the state; the track is its visual representation.
  const notificationBugfixToggle = document.getElementById("notifications-bugfix");
  const notificationBugfixStatus = document.getElementById("notifications-bugfix-status");
  notificationBugfixToggle.addEventListener("change", () => {
    notificationBugfixStatus.textContent = notificationBugfixToggle.checked
      ? "Notifications are on."
      : "Notifications are off.";
  });
</script>
```

## Homework — Interactive PVO Toggle Panel

The three switches each control a visible panel, and the live total reports how many settings are on. The daily reminder begins enabled; change any switch to test both states. Comments explain the `.checked` state and how the count is updated.

```python
%%html
<!-- UI_RUNNER: Homework - three interactive PVO preference toggles -->
<div id="pvo-toggle-homework">
  <h3>PVO Support Preferences</h3>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="pvo-daily-reminders" class="ocs__toggle-input" type="checkbox" checked>
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Daily wellness reminders</span>
  </label>
  <p id="pvo-reminder-panel">Daily reminders are enabled. Choose a time that works for you.</p>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="pvo-breathing-tips" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show breathing-break tips</span>
  </label>
  <p id="pvo-breathing-panel" hidden>Try a slow breath in, then a longer breath out.</p>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="pvo-volunteer-followup" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Allow volunteer follow-up</span>
  </label>
  <p id="pvo-followup-panel" hidden>A PVO volunteer can follow up about the resources you requested.</p>

  <output id="pvo-toggle-count" aria-live="polite">1 of 3 settings enabled.</output>
</div>
<script>
  const pvoDailyReminders = document.getElementById("pvo-daily-reminders");
  const pvoBreathingTips = document.getElementById("pvo-breathing-tips");
  const pvoVolunteerFollowup = document.getElementById("pvo-volunteer-followup");
  const pvoToggleCount = document.getElementById("pvo-toggle-count");

  // Each checkbox's checked property is the true/false state of that preference.
  function updatePvoTogglePanel() {
    document.getElementById("pvo-reminder-panel").hidden = !pvoDailyReminders.checked;
    document.getElementById("pvo-breathing-panel").hidden = !pvoBreathingTips.checked;
    document.getElementById("pvo-followup-panel").hidden = !pvoVolunteerFollowup.checked;

    // Count enabled settings so the summary always matches the switches.
    const enabledCount = [pvoDailyReminders, pvoBreathingTips, pvoVolunteerFollowup]
      .filter((toggle) => toggle.checked).length;
    pvoToggleCount.textContent = enabledCount + " of 3 settings enabled.";
  }

  [pvoDailyReminders, pvoBreathingTips, pvoVolunteerFollowup].forEach((toggle) => {
    toggle.addEventListener("change", updatePvoTogglePanel);
  });
  updatePvoTogglePanel();
</script>
```
