# Semantic HTML with OCS Typography Grammar

Completed the semantic refactor Popcorn Hack and the cat-care instructions homework using meaningful HTML and OCS typography classes.

## Popcorn Hack — Refactor “Why Cats Purr”

**Completed:** Replaced visual-only `div`/`span` elements and custom styling with a semantic heading hierarchy and paragraph. The supporting paragraph uses the OCS `ocs_lead` role.

```python
%%html
<!-- UI_RUNNER: Typography Popcorn - semantic refactor -->
<h1>Why Cats Purr</h1>
<h2>A Sign of Happiness</h2>
<p class="ocs_lead">Cats usually purr when they feel safe, cozy, and ready for a nap.</p>
```

## Homework Hack — How to Pat a Cat

**Completed:** The refactor uses an OCS card with a section title, a heading hierarchy, paragraphs, semantic `<strong>` emphasis, and a real ordered list of three steps. Custom CSS and invented classes have been removed.

```python
%%html
<!-- UI_RUNNER: Typography Homework Base -->
<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <h3>Have you ever wanted to...</h3>
  <p>Pat a cat? Well, good for you! Here are step-by-step instructions on how to pat a cat.</p>
  <p><strong>How to pat a cat:</strong></p>
  <ol>
    <li>Put out your hand.</li>
    <li>Touch the cat gently, if it welcomes the contact.</li>
    <li>Move your hand back and forth slowly.</li>
  </ol>
  <p>Congratulations! You patted a cat.</p>
</div>
```
