```
DISPLAY("Grade Entry System")

grades ← [] 
gradeCount ← 0
totalPoints ← 0
continueEntry ← "yes"

REPEAT UNTIL (continueEntry = "no")
{
  grade ← INPUT("Add a grade:")
  
  APPEND(grades, grade)
  gradeCount ← gradeCount + 1
  totalPoints ← totalPoints + grade
  
  continueEntry ← INPUT("Continue entering grades? (no to stop)")
}

DISPLAY("List of grades: " + grades)
DISPLAY("Total grades entered: " + gradeCount)
DISPLAY("Total points: " + totalPoints)
```

````python
print("Digital Checkout")

items = []
itemCount = 0
totalPrice = 0
continueEntry = True

while continueEntry:
    item = input("Enter the item you want to buy: ")

    items.append(item)
    itemCount += 1
    totalPrice = totalPrice + item

    continueEntry = input("Do you want to enter another item? (y/n): ")

print("You have entered " + str(itemCount) + " items.")
print("The total price is: " + str(totalPrice))
````
